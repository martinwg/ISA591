# 11. Why OOP Matters for Machine Learning

Machine-learning libraries use the same object-oriented pattern.

A typical workflow is:

1. **Instantiate** a model object.
2. Call `.fit()` to train it.
3. Call `.predict()` to generate predictions.

Example:

```python
model = LogisticRegression()
model.fit(X, y)
predictions = model.predict(X_test)
```

`LogisticRegression` is a class.  
`model` is an instance of that class.  
`.fit()` and `.predict()` are methods.
