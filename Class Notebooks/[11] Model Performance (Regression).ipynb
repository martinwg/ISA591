# 1. What Is a Good Prediction?

For observation $i$:

$$e_i=y_i-\hat{y}_i$$

A good model produces residuals close to zero.

Positive residual → model **underpredicted**.  
Negative residual → model **overpredicted**.

## Practice 1A — Calculate residuals by hand

```python
## IDEALLY - SPLIT THE DATA INTO TRAINING AND TEST
#### build models on TRAINING
#### evaluate the models on TEST

example=pd.DataFrame({
    "Actual":[15000,22000,18000,30000],
    "Predicted":[14000,25000,17500,27000]
})

# TODO
example["Residual"]= example['Actual'] - example['Predicted']
example
```

```python
## 1) MEAN SQUARED ERROR (MSE)

#### drawback 1: squared error can be impacted by outliers
#### drawback 2: squared units (i.e., dollars squared) are not interpretable
example['Squared Error'] = example['Residual']**2
example['Squared Error'].mean()
```

```python
## 2) ROOT MEAN SQUARED ERROR (RMSE)

### Advantages: interpretable (The estimated error of the model is $2194)
### Drawback: still impacted by outliers
np.sqrt(example['Squared Error'].mean())
```

```python
## 3) MEAN ABSOLUTE ERROR (MAE)

### Advantages: interpretable (The estimated error of the model)
### Advantages: not as impacted by outliers
example['Absolute Error'] = np.abs(example['Residual'])
example['Absolute Error'].mean()
```

# 3. Regression Performance Metrics

Different metrics summarize prediction error differently.

### MSE
$$MSE=\frac{1}{n}\sum(y_i-\hat y_i)^2$$

Lower is better. Squaring heavily penalizes large errors, but the units are squared.

### RMSE
$$RMSE=\sqrt{MSE}$$

Lower is better and it returns to the original units of $y$.

### MAE
$$MAE=\frac{1}{n}\sum|y_i-\hat y_i|$$

Lower is better and is less sensitive to extreme errors than RMSE.

### MAPE
$$MAPE=\frac{1}{n}\sum\left|\frac{y_i-\hat y_i}{y_i}\right|$$

Expresses error relative to the actual value. It becomes problematic when actual values are zero or very close to zero.

### $R^2$
Measures improvement relative to predicting the mean. Higher is generally better; test-set $R^2$ can be negative.

# 5. Residual Diagnostics

A single average metric can hide unusual errors.

Plot residuals and inspect their distribution.

```python
# TODO
residuals=y_test-y_pred_rf

fig,ax=plt.subplots(1,2,figsize=(12,4))
ax[0].scatter(y_pred_rf,residuals,alpha=.4)
ax[0].axhline(0,linestyle="--")
ax[0].set_xlabel("Predicted Selling Price")
ax[0].set_ylabel("Residual")
ax[0].set_title("Residuals vs Predictions")

ax[1].hist(residuals,bins=30)
ax[1].set_title("Residual Distribution")
plt.show()
```
